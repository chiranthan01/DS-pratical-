# Practical 1: Matrix Operations and Encoding

**Dataset:** StudentsPerformance.csv

**Aim:** Load the data, convert scores to NumPy arrays, perform matrix operations and apply one-hot encoding.

```python
import pandas as pd
import numpy as np

students = pd.read_csv("StudentsPerformance.csv")
students.head()
```

**Result:** The dataset has 1000 rows and 8 columns.

```python
students.info()
```

**Result:** All columns have 1000 non-null values, so there are no missing values. Three columns are numeric (scores) and five are categorical.

```python
scores = students[["math score", "reading score", "writing score"]].to_numpy()
print("Shape     :", scores.shape)
print("Size      :", scores.size)
print("Dimensions:", scores.ndim)
```

**Result:** The score array has shape (1000, 3), size 3000 and 2 dimensions.

```python
part = students.loc[0:3, ["math score", "reading score"]]
print(part)
```

**Result:** The first four students with their math and reading scores were selected.

```python
values = part.to_numpy()
A = values[0:2]
B = values[2:4]
print("A =")
print(A)
print("B =")
print(B)
```

**Result:** The 4 x 2 array was split into two 2 x 2 matrices, A and B.

```python
total = np.add(A, B)
print(total)
```

**Result:** A + B = [[162, 167], [116, 147]]

```python
difference = np.subtract(A, B)
print(difference)
```

**Result:** A - B = [[-18, -23], [22, 33]]

```python
elementwise = np.multiply(A, B)
print(elementwise)
```

**Result:** Element-wise product of A and B = [[6480, 6840], [3243, 5130]]

```python
product = np.matmul(A, B)
print(product)
```

**Result:** Matrix product A x B = [[9864, 10944], [10440, 11685]]

```python
determinant = np.linalg.det(A)
inverse = np.linalg.inv(A)
print("Determinant:", round(determinant, 2))
print(inverse)
print(np.round(A @ inverse, 6))
```

**Result:** The determinant of A is 1512.00, which is not zero, so the inverse exists. Multiplying A by its inverse gives the identity matrix, which confirms the inverse is correct.

```python
transpose = A.transpose()
print(transpose)
```

**Result:** Transpose of A = [[72, 69], [72, 90]]

```python
text_columns = [c for c in students.columns if not pd.api.types.is_numeric_dtype(students[c])]
encoded = pd.get_dummies(students, columns=text_columns, dtype=int)
print("Categorical columns:", text_columns)
print("Shape before:", students.shape)
print("Shape after :", encoded.shape)
encoded.head()
```

**Result:** The five categorical columns were converted into 17 binary columns. The table grew from 8 to 20 columns.
