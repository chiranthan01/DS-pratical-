# Practical 5: Probability and Hypothesis Testing

**Dataset:** TeachingRatings.csv

**Aim:** Find probabilities using the normal distribution and perform a two-tailed z-test.

```python
import pandas as pd
from scipy import stats

ratings = pd.read_csv("TeachingRatings.csv")
mu = ratings["eval"].mean()
sd = ratings["eval"].std()
print("Mean:", round(mu, 4))
print("Std :", round(sd, 4))
```

**Result:** The evaluation score has mean 3.9983 and standard deviation 0.5549. These are used as the normal distribution parameters.

### Question 1: P(evaluation score > 4.5)

```python
z1 = (4.5 - mu) / sd
p1 = stats.norm.sf(z1)
actual1 = (ratings["eval"] > 4.5).mean()
print("z-score         :", round(z1, 4))
print("Normal model    :", round(p1, 4))
print("Actual in data  :", round(actual1, 4))
```

**Result:** z = 0.9042. The probability of a score above 4.5 is 0.1829 (18.29%) from the normal model, and 17.06% in the actual data.

### Question 2: P(3.5 < evaluation score < 4.2)

```python
za = (3.5 - mu) / sd
zb = (4.2 - mu) / sd
p2 = stats.norm.cdf(zb) - stats.norm.cdf(za)
actual2 = ratings["eval"].between(3.5, 4.2, inclusive="neither").mean()
print("z-scores        :", round(za, 4), round(zb, 4))
print("Normal model    :", round(p2, 4))
print("Actual in data  :", round(actual2, 4))
```

**Result:** The probability of a score between 3.5 and 4.2 is 0.4573 (45.73%) from the normal model, and 35.42% in the actual data. The model gives a higher value because scores are not perfectly normal.

### Question 3: Two-tailed z-test

A basketball team's sample is compared with the regional players. Known population mean = 12, population standard deviation = 5.5, sample size = 36, sample mean = 10.7.

H0: the team mean equals 12. H1: the team mean is different from 12. Significance level = 0.05.

```python
pop_mean, pop_sd, n, sample_mean, alpha = 12, 5.5, 36, 10.7, 0.05
std_error = pop_sd / n ** 0.5
z = (sample_mean - pop_mean) / std_error
p_value = 2 * stats.norm.sf(abs(z))
z_crit = stats.norm.ppf(1 - alpha / 2)
print("Standard error :", round(std_error, 4))
print("z statistic    :", round(z, 4))
print("Critical value :", round(z_crit, 4))
print("p-value        :", round(p_value, 4))
verdict = "reject H0" if abs(z) > z_crit else "fail to reject H0"
print("Decision       :", verdict)
```

**Result:** z = -1.4182 lies inside the range -1.96 to +1.96 and the p-value 0.1561 is greater than 0.05, so we fail to reject H0. There is no significant difference between the team mean and the regional mean.
