Step 1: Creation and Loading of Data

Firstly, we generate a large data set for an e-commerce company using Python libraries Pandas and Numpy. The data set comprises 50,000 entries that include information like Order ID, Product, Quantity, Price, and City. For random number generation, Numpy library is used whereas for conversion into DataFrame, Pandas is used. Further, head() function is used to display top 5 entries and shape to identify number of rows and columns.

Step 2: Cleaning and Preparation of Data

Secondly, to check whether there are any missing values in the data set, we use isnull().sum() function. This helps in identifying if any of the data is missing in the data set. Then, we create another column by the name of Total_Sales as multiplication of Quantity and Price. This process is known as data preprocessing as here we make data ready to analyze.

Step 3: Analyze and Group the Data

In the third step, we analyze the data by applying the groupby() function. We group the sales based on Product and get the total sales of the products using the sum(). We can even group the data based on City to know the total sales in each city. By this way, we get to know the better-performing products and locations.

Step 4: Perform Statistical Analysis

In the fourth step, we will apply the statistical analysis to the Total_Sales column using NumPy. We will calculate the mean, median, minimum, maximum, and standard deviation of Total_Sales. Mean represents the average sales, the median is the middle value of sales, the minimum and maximum represent the least and greatest values of sales, and standard deviation represents the variation of the sales values from the mean.

Step 5: Visualize the Data

In the last step, we will visualize the data by creating a histogram for Total_Sales using Matplotlib. By creating a histogram, we will know whether the sales values are distributed around the certain values or are widely scattered or are skewed.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python


np.random.seed(10)

data = {
    'Order_ID': range(1, 50001),
    'Product': np.random.choice(
        ['Laptop', 'Mobile', 'Tablet', 'Headphones'], 50000
    ),
    'Quantity': np.random.randint(1, 10, 50000),
    'Price': np.random.randint(500, 80000, 50000),
    'City': np.random.choice(
        ['Delhi', 'Mumbai', 'Pune', 'Chandigarh'], 50000
    )
}

df = pd.DataFrame(data)

print(df.head())
print("Shape:", df.shape)
```

```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Create total sales
df['Total_Sales'] = df['Quantity'] * df['Price']

print("\nUpdated Data:")
print(df.head())
```

```python
# Total sales by product
product_sales = df.groupby('Product')['Total_Sales'].sum()

print("Sales by Product:")
print(product_sales)

# Total sales by city
city_sales = df.groupby('City')['Total_Sales'].sum()

print("\nSales by City:")
print(city_sales)
```

```python
sales = df['Total_Sales'].to_numpy()

print("Mean Sales:", np.mean(sales))
print("Median Sales:", np.median(sales))
print("Minimum Sales:", np.min(sales))
print("Maximum Sales:", np.max(sales))
print("Standard Deviation:", np.std(sales))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df['Total_Sales'], bins=30, edgecolor='black')

plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.title("Distribution of Total Sales")

plt.show()
```
